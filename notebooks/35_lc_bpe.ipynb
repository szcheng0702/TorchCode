# 🔴 Hard: Byte-Pair Encoding (BPE)

Implement a simple **BPE tokenizer** — the foundation of GPT/LLaMA tokenization.

### Signature
```python
class SimpleBPE:
    def __init__(self): ...
    def train(self, corpus: list[str], num_merges: int): ...
    def encode(self, text: str) -> list[str]: ...
```

### Algorithm (training)
1. Split each word into characters + `</w>` end marker
2. Count all adjacent pairs across the corpus
3. Merge the most frequent pair into a single token
4. Repeat for `num_merges` iterations

````python
# No imports needed
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

class SimpleBPE:
    def __init__(self):
        self.merges = []

    def train(self, corpus, num_merges):
        vocab = dict()
        for word in corpus:
            s = tuple(word)+('</w>',)
            vocab[s] = vocab.get(s,0)+1
        for _ in range(num_merges):
            pairs = dict()
            for w,freq in vocab.items():
                for i in range(len(w)-1):
                    pair = w[i], w[i+1]
                    pairs[pair] = pairs.get(pair,0)+freq\
            if not pairs:
                    break
            best_pair = max(pairs,key=pairs.get)
            self.merges.append(best_pair)
            # update vocab
            new_vocab = dict()
            for w,freq in vocab.items():
                new_word = []
                i = 0
                while i < len(w):
                    if i < len(w) - 1 and (w[i], w[i + 1]) == best_pair:
                        new_word.append(w[i] + w[i + 1])
                        i += 2
                    else:
                        new_word.append(w[i])
                        i += 1
                new_vocab[tuple(new_word)]=freq
            vocab = new_vocab

    def encode(self, text):
        tokens=[]
        for word in text.split():
            s = tuple(word)+('</w>',)
            for a,b in self.merges:
                i = 0
                while i<len(s)-1:
                    if s[i] == a and s[i+1] == b:
                        s = s[:i]+(a+b,)+s[i+2:]
                    i+=1
            tokens.append(s)
        return tokens
                
                
````

````python
# 🧪 Debug
bpe = SimpleBPE()
bpe.train(['low', 'low', 'low', 'lower', 'newest', 'widest'], num_merges=10)
print('Merges:', bpe.merges[:5])
print('Encode:', bpe.encode('low lower'))
````
